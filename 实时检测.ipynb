```python
import cv2

from ultralytics import YOLO


def detection(model_path, camera_id=1, conf_thre=0.7):
    """
    Args:
        model_path: 权重文件
        camera_id: 摄像头ID
        conf_thre:置信度阈值.

    Returns:
    """
    # 加载yolov8模型
    model = YOLO(model_path)
    # 打开摄像头
    cap = cv2.VideoCapture(camera_id)
    cap.set(cv2.CAP_PROP_FRAME_WIDTH, 640)  # 设置画面宽度
    cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 480)  # 设置画面高度
    cap.set(cv2.CAP_PROP_FPS, 30)  # 设置帧率
    print("开始检测，按’q‘退出")
    while True:
        _ret, frame = cap.read()
        # 执行检测
        results = model(frame, conf=conf_thre)
        for results in results:
            boxes = results.boxes
            for box in boxes:
                x1, y1, x2, y2 = map(int, box.xyxy[0])
                conf = float(box.conf[0])
                cls = int(box.cls[0])
                # 类别字典
                # {0:"person",1:"car",2:"bicyle"}
                cls_name = model.names[cls]
                # 绘制矩形框
                cv2.rectangle(frame, (x1, y1), (x2, y2), (255, 255, 0), 2)
                # 增加标签文字
                # 文字格式：person 0.85
                label = f"{cls_name} {conf:.2f}"
                cv2.putText(frame, label, (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255, 255, 255), 1)
        cv2.imshow("YOLO", frame)

        # 设置按键退出
        key = cv2.waitKey(1) & 0xFF
        if key == ord("q"):
            break
    cap.release
    cap.destoryAllWindows()


if __name__ == "__main__":
    model_path = r"C:\Users\31440\Desktop\yolov8\yolo11n.pt"
    camera_id = 1
    conf_thre = 0.5

    detection(model_path, camera_id, conf_thre)
```
