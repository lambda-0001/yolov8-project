```python
import os

import cv2

mp4 = "./test1.mp4"
dir_path = "./test1_images/"


video = cv2.VideoCapture(mp4)  # 使用opencv读取视频流
success, images = video.read()

fps = int(video.get(cv2.CAP_PROP_FPS))  # 视频帧率1秒30张
count = 0

if not os.path.exists(dir_path):
    os.mkdir(dir_path)

while success:
    if count % fps == 0 and count != 0:  # 每1秒截取1帧
        cv2.imwrite(f"{dir_path}/{int(count / fps)}.jpg", images)
        print(f"已保存第{int(count / fps)}张图片")
    success, images = video.read()
    count += 1
```
