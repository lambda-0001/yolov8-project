```python
import cv2
import matplotlib.pyplot as plt
import xml.etree.ElementTree as ET
image = cv2.imread(r'C:\Users\31440\Desktop\yolov8\datasets\images\train\1.jpg')  # 读取图片
label = open(r"D:\yolov5\datasets\seer\labels\1.xml", encoding="utf-8")  # 读取标签
# 解析xml文件
tree = ET.parse(r"D:\yolov5\datasets\seer\labels\1.xml")
root = tree.getroot()
# 获得图片的尺寸大小
for obj in root.iter("object"):
    cls = obj.find("name").text
    xmlbox = obj.find("bndbox")
    print(cls)
    xmin = int(xmlbox.find("xmin").text)
    ymin = int(xmlbox.find("ymin").text)
    xmax = int(xmlbox.find("xmax").text)
    ymax = int(xmlbox.find("ymax").text)
    cv2.rectangle(image, (xmin,ymin),(xmax,ymax),(0,0,255),1)
    cv2.putText(image,cls, (xmin,ymin+10),cv2.FONT_HERSHEY_SIMPLEX, 0.5,(0,0,255))
    img_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(12,8))
plt.imshow(img_rgb)
plt.show()
```
